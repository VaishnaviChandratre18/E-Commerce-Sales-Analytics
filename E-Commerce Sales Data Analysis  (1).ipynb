```python
import pandas as pd
import numpy as np

Mean=np.mean(df['Revenue'])   #Mean
print(Mean)
```

```python
#Maximum Revenue

Maximum=np.max(df['Revenue'])
print(Maximum)

Row=df[df['Revenue']==df['Revenue'].max()]
print(Row)       #  Headphones has the maximum reveune with 430754.76 on 21-10-2023
```

```python
# Minimum Revenue

Minimum=np.min(df['Revenue'])
print(Minimum)

Minrow=df[df['Revenue']==df['Revenue'].min()]
print(Minrow)              # Smart watch has generated least revenue of 7495.02 on 25-8-2023
```

```python
# Top selling product

Prod=np.max(df['Revenue'])
da=df[df['Revenue']==df['Revenue'].max()]
print(da)              #Headphones is the top selling product
```

```python
# TOP 5 selling products

top_products=df.groupby('Product')['Revenue'].sum().sort_values(ascending=False).head(5)
print(top_products)
```

```python
# Least 5 selling product

least=df.groupby('Product')['Revenue'].sum().sort_values(ascending=True).head(5)
print(least)
```

```python
#Monthly revenue

import pandas as pd
import matplotlib.pyplot as plt

dp=df.groupby('Month')['Revenue'].sum()

plt.plot(dp.index,dp.values,linestyle='--',marker='o',color='deeppink')
plt.title("Monthly revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.savefig("Monthly revenue trend.png")
plt.show()
```

```python
#Category wise revenue
import pandas as pd
import matplotlib.pyplot as plt
plt.bar(df['Category'],df['Revenue'],color='pink')
plt.xlabel("Categories")
plt.ylabel("Revenue")
plt.title("Category wise revenue")
plt.savefig("Category wise revenue.png")
plt.show()
```

```python
#Product wise Revenue
import pandas as pd
import matplotlib.pyplot as plt
plt.barh(df['Product'],df['Revenue'],color='palevioletred')
plt.xlabel("Revenue")
plt.ylabel("Product")
plt.title("Product wise Revenue")
plt.savefig("Product wise Revenue.png")
plt.show()
```

```python
#Correlation
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
corr=df[['Quantity','Price','Revenue']].corr()
sns.heatmap(corr,annot=True,cmap='PuRd',vmin=-1,vmax=1)
plt.savefig("Correlation.png")
plt.show()
```

```python
# Revenue Distribution
import pandas as pd
import matplotlib.pyplot as plt
plt.hist(df['Revenue'],color='palevioletred')
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.title("Revenue distribution")
plt.savefig("Revenue Distribution.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
plt.boxplot(df['Revenue'])
plt.ylabel("Revenue")
plt.title("Revenue Boxplot")
plt.savefig("No outliers")
plt.show()      #No outliers in revenue
```
